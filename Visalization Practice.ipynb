```python
#calculating correlation 
```

```python
correlation_cols=["Length","Width","Height"]
corr_matrix=df[correlation_cols].corr()
```

```python
df.columns
```

```python
print("Correlation Matrix:")
print(corr_matrix.round(2).to_string())
```

```python
columns=["Length","Width","Height"]
corr=df[columns].corr()
```

```python
print("Correlation:")
print(corr.round(1).to_string())
```

```python
print("Correlation:")
print(corr.round(5).to_string())
```

```python
# create correlation Heatmap
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
sns.heatmap(corr_matrix, annot=True, cmap='RdYlBu_r', center=0, 
            fmt='.2f', square=True, linewidths=0.5,
            annot_kws={'size': 10})
plt.title("Heatmap for df", fontsize=14, fontweight='bold')
plt.tight_layout()

plt.show()
```

```python
fig,ax=plt.subplots(figsize=(6,4))
sns.heatmap(corr,annot=True,cmap="RdYlGn",center=0,fmt=".2g",square=True,linewidths=0.5, annot_kws={"size":10})

plt.title("Heatmap for df", fontweight="bold")
plt.tight_layout()
plt.show()
```
